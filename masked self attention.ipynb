# Function Signature

```python
import numpy as np
import math

def causal_mask(seq_len):
    causal_mask = np.triu(
        np.full((seq_len, seq_len), -np.inf),
        k=1
    )
    return causal_mask


def softmax(x: np.ndarray, axis: int = -1) -> np.ndarray:
    x = x - np.max(x, axis=axis, keepdims=True)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x, axis=axis, keepdims=True)


def causal_self_attention(Q, K, V):

    batch, seq_len, d_model = Q.shape

    d_k = K.shape[-1]

    # Step 1: QKᵀ
    qk = Q @ K.transpose(0, 2, 1)

    # Step 2: Scale
    scores = qk / math.sqrt(d_k)

    # Step 3: Causal mask
    causal = causal_mask(seq_len)

    # Step 4: Apply mask
    masked = scores + causal

    # Step 5: Softmax
    attention = softmax(masked, axis=-1)

    # Step 6: Attention × V
    output = attention @ V

    return output, attention
```

# code run

```python
Q = np.array([[
    [1.0, 0.0],
    [0.0, 1.0],
    [1.0, 1.0]
]])

K = np.array([[
    [1.0, 0.0],
    [0.0, 1.0],
    [1.0, 1.0]
]])

V = np.array([[
    [10.0, 0.0],
    [0.0, 20.0],
    [30.0, 30.0]
]])

output, attention = causal_self_attention(Q, K, V)

print("Attention:")
print(attention)

print("\nOutput:")
print(output)
```
