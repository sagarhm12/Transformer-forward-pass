```python
def build_vocab(texts):
    
    text_lower=[text.lower() for text in texts]
    vocab_set = set(text_lower)
    special_tokens = ["<PAD>", "<UNK>", "<SOS>", "<EOS>"]
    word_to_id={token:idx for idx,token in enumerate(special_tokens)}
    for idx,word in enumerate(sorted(vocab_set),start=len(special_tokens)):
        word_to_id[word]=idx

    return word_to_id
```

```python
texts=['hello','unknown','sagar','hi','hi']
```

```python

vocab=build_vocab(texts)
```

```python
def encode(text: str) -> list[int]:
    text=text.lower()
    words=text.split()
    encoder=[]
    for word in words:
        if word not in vocab:
            encoder.append(1)
        else:
            encoder.append(vocab[word])

    return encoder
    
```

```python
ar=encode('hi my name is sagar')
```
